```python
"""
Interactive File Handling System in Python
============================================
A menu-driven program covering every basic file-handling concept.
The filename and content are taken from the USER at runtime

Concepts covered:
  1. CREATE a file
  2. WRITE to a file (overwrite)
  3. READ a file (whole content)
  4. READ line by line
  5. APPEND to a file
  6. CHECK if a file exists
  7. RENAME a file
  8. DELETE a file
  9. Exception handling around file operations
"""

import os


def create_file():
    filename = input("Enter filename to create: ").strip()
    if os.path.exists(filename):
        print(f"⚠️  '{filename}' already exists — nothing created.\n")
        return
    try:
        with open(filename, "x"):
            pass
        print(f"✅ '{filename}' created (empty).\n")
    except OSError as e:
        print(f"❌ Could not create file: {e}\n")


def write_file():
    filename = input("Enter filename to write to: ").strip()
    print("Enter lines of text. Type 'end' on its own line to stop.")
    lines = []
    while True:
        line = input("> ")
        if line.strip().lower() == "end":
            break
        lines.append(line + "\n")

    try:
        with open(filename, "w") as f:
            f.writelines(lines)
        print(f"✅ Wrote {len(lines)} line(s) to '{filename}' (previous content overwritten).\n")
    except OSError as e:
        print(f"❌ Could not write file: {e}\n")


def read_file():
    filename = input("Enter filename to read: ").strip()
    try:
        with open(filename, "r") as f:
            content = f.read()
        print(f"\n--- Content of '{filename}' ---")
        print(content if content else "(file is empty)")
        print("--- End ---\n")
    except FileNotFoundError:
        print(f"❌ '{filename}' does not exist.\n")


def read_lines():
    filename = input("Enter filename to read line by line: ").strip()
    try:
        with open(filename, "r") as f:
            lines = f.readlines()
        if not lines:
            print("(file is empty)\n")
            return
        for i, line in enumerate(lines, start=1):
            print(f"  {i}: {line.rstrip()}")
        print()
    except FileNotFoundError:
        print(f"❌ '{filename}' does not exist.\n")


def append_file():
    filename = input("Enter filename to append to: ").strip()
    print("Enter lines to append. Type 'done' on its own line to stop.")
    lines = []
    while True:
        line = input("> ")
        if line.strip().lower() == "done":
            break
        lines.append(line + "\n")

    try:
        # "a" mode creates the file automatically if it doesn't exist
        with open(filename, "a") as f:
            f.writelines(lines)
        print(f"✅ Appended {len(lines)} line(s) to '{filename}'.\n")
    except OSError as e:
        print(f"❌ Could not append to file: {e}\n")


def check_exists():
    filename = input("Enter filename to check: ").strip()
    if os.path.exists(filename):
        print(f"✅ '{filename}' exists. Is a file: {os.path.isfile(filename)}\n")
    else:
        print(f"❌ '{filename}' does not exist.\n")


def rename_file():
    old_name = input("Enter the current filename: ").strip()
    if not os.path.exists(old_name):
        print(f"❌ '{old_name}' does not exist.\n")
        return
    new_name = input("Enter the new filename: ").strip()
    if os.path.exists(new_name):
        print(f"⚠️  '{new_name}' already exists — rename aborted.\n")
        return
    try:
        os.rename(old_name, new_name)
        print(f"✅ Renamed '{old_name}' -> '{new_name}'.\n")
    except OSError as e:
        print(f"❌ Could not rename file: {e}\n")


def delete_file():
    filename = input("Enter filename to delete: ").strip()
    if not os.path.exists(filename):
        print(f"❌ '{filename}' does not exist — nothing to delete.\n")
        return
    confirm = input(f"Are you sure you want to delete '{filename}'? (y/n): ").strip().lower()
    if confirm != "y":
        print("Delete cancelled.\n")
        return
    try:
        os.remove(filename)
        print(f"✅ '{filename}' deleted.\n")
    except OSError as e:
        print(f"❌ Could not delete file: {e}\n")


def menu():
    options = {
        "1": ("Create a file", create_file),
        "2": ("Write to a file (overwrite)", write_file),
        "3": ("Read a file (full content)", read_file),
        "4": ("Read a file (line by line)", read_lines),
        "5": ("Append to a file", append_file),
        "6": ("Check if a file exists", check_exists),
        "7": ("Rename a file", rename_file),
        "8": ("Delete a file", delete_file),
        "9": ("Exit", None),
    }

    while True:
        print("===== File Handling System =====")
        for key, (label, _) in options.items():
            print(f"{key}. {label}")
        choice = input("Enter choice (1-9): ").strip()

        if choice == "9":
            print("Goodbye!")
            break
        elif choice in options:
            options[choice][1]()
        else:
            print("⚠️  Invalid choice, try again.\n")


if __name__ == "__main__":
    menu()
```
